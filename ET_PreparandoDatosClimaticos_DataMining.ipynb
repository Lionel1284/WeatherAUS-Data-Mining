#### 3.6.2 Correlaciones

Para comprender cómo interactúan las variables numéricas de nuestro conjunto de datos, utilizaremos una **Matriz de Correlación de Pearson**. Este análisis mide la relación lineal entre dos variables, arrojando un valor entre -1 y 1:
* **Cercano a 1:** Correlación positiva fuerte (si una variable aumenta, la otra también).
* **Cercano a -1:** Correlación negativa fuerte (si una variable aumenta, la otra disminuye).
* **Cercano a 0:** No hay relación lineal aparente.

**Objetivos de este análisis:**
1. **Identificar Multicolinealidad:** Si dos variables predictoras están altamente correlacionadas entre sí (ej. `Temp9am` y `MaxTemp`), podrían aportar información redundante al modelo. Algoritmos como la Regresión Lineal y Logística sufren problemas de inestabilidad ante la multicolinealidad.
2. **Descubrir Patrones Climáticos:** Entender dinámicas naturales, como la relación inversa que suele existir entre la humedad y la temperatura, o cómo la presión atmosférica interactúa con las precipitaciones.

A continuación, generaremos un mapa de calor (*Heatmap*) para visualizar estas relaciones de forma rápida e intuitiva.

```python
df_numerico = df.select_dtypes(include=['float64', 'int64'])

matriz_correlacion = df_numerico.corr()

plt.figure(figsize=(18, 14))

mascara = np.triu(np.ones_like(matriz_correlacion, dtype=bool))

sns.heatmap(matriz_correlacion,
            mask=mascara,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            vmin=-1, vmax=1,
            linewidths=0.5,
            cbar_kws={"shrink": .8})

plt.title('Mapa de Calor de Correlaciones de Pearson (Variables Numéricas)', fontsize=18, pad=20)
plt.xticks(rotation=45, ha='right', fontsize=11)
plt.yticks(fontsize=11)
plt.tight_layout()

plt.show()
```

El mapa de calor revela relaciones lineales muy marcadas entre diversas variables meteorológicas. Dentro del marco de la **Fase 2: Comprensión de los Datos**, el análisis de estos coeficientes nos permite entender la estructura interna del dataset y cómo interactúan las variables entre sí:

**1. Identificación de Redundancia Lineal Alta (Multicolinealidad)**
Se detectan varios pares de variables que registran un comportamiento lineal prácticamente idéntico, lo que significa que aportan la misma información climática al conjunto de datos:
* **`MaxTemp` y `Temp3pm` (r = 0.98):** Existe una relación casi perfecta, indicando que la temperatura máxima diaria está fuertemente determinada por la medición de la tarde.
* **`Pressure9am` y `Pressure3pm` (r = 0.96):** La presión atmosférica se mantiene muy estable entre la mañana y la tarde, mostrando una variación lineal mínima.
* **`MinTemp` y `Temp9am` (r = 0.90):** La temperatura de la mañana está estrechamente ligada al comportamiento de la temperatura mínima registrada.

**2. Relaciones Climáticas Inversas**

### 4.2 Tratamiento e Imputación de Valores Nulos

Una vez descartadas las variables inviables por su alto volumen de vacíos, el dataset actual aún conserva variables con porcentajes menores de datos faltantes (menores al 10%). Dejar estos nulos impediría el correcto funcionamiento de los algoritmos de Machine Learning, ya que la mayoría de ellos no aceptan entradas vacías.

Para solucionar esto de manera estadísticamente rigurosa y alineada con los hallazgos de nuestra "**Fase 2 Data Understanding**", aplicaremos dos estrategias diferenciadas según la naturaleza de la variable:

1. **Variables Numéricas (Imputación por Mediana):** En lugar de utilizar la media aritmética, utilizaremos la **mediana** para rellenar los valores faltantes en variables como `Rainfall`, `WindGustSpeed`, `Humidity3pm`, etc. Como demostramos en los histogramas y boxplots del EDA, muchas de estas variables sufren de una fuerte asimetría y outliers severos; la mediana es una medida de tendencia central robusta que no se ve afectada por estos extremos, protegiendo la distribución original.
   
2. **Variables Categóricas (Imputación por Moda):**
   Para las variables de texto o cualitativas (como `WindGustDir`, `WindDir9am`, `WindDir3pm` y `RainToday`), no es posible calcular un centro numérico. En su lugar, utilizaremos la **moda** (el valor o categoría más frecuente en el histórico) para rellenar los espacios vacíos.

```python
print("--- ESTADO INICIAL DE NULOS EN EL DATASET ---")
print(df_preparado.isnull().sum()[df_preparado.isnull().sum() > 0])
print("-" * 50)

columnas_numericas = df_preparado.select_dtypes(include=['float64', 'int64']).columns
columnas_categoricas = df_preparado.select_dtypes(include=['object']).columns

if 'Date' in columnas_categoricas:
    columnas_categoricas = columnas_categoricas.drop('Date')

for col in columnas_numericas:
    mediana_col = df_preparado[col].median()
    df_preparado[col] = df_preparado[col].fillna(mediana_col)

for col in columnas_categoricas:
    moda_col = df_preparado[col].mode()[0]
    df_preparado[col] = df_preparado[col].fillna(moda_col)

print("\n--- ESTADO FINAL DE NULOS TRAS LA IMPUTACIÓN ---")
nulos_restantes = df_preparado.isnull().sum().sum()
print(f"Cantidad total de valores nulos remanentes en el dataset: {nulos_restantes}")

if nulos_restantes == 0:
    print("¡Éxito! El conjunto de datos ya no contiene valores nulos y está listo para los siguientes pasos.")
```

La ejecución del proceso de imputación ha logrado restaurar la completitud del conjunto de datos al 100%, dejándolo estructuralmente apto para la fase de modelado. El impacto analítico de este tratamiento se resume en los siguientes puntos:

1. **Recuperación de la integridad sin pérdida adicional de datos:** En lugar de eliminar filas por cada valor nulo encontrado, se conservan los registros disponibles después de remover únicamente aquellos casos sin variable objetivo `RainTomorrow`. Esto evita una pérdida innecesaria de información meteorológica.
2. **Preservación de la distribución original:** Al utilizar la **mediana** para las variables numéricas y la **moda** para las categóricas, se rellenan los espacios vacíos sin alterar de forma agresiva el centro de la distribución. Esta decisión protege la estructura general del clima australiano y reduce el impacto que los valores extremos habrían tenido si se utilizara la media aritmética.
3. **Compatibilidad con algoritmos de Machine Learning:** La mayoría de los modelos estadísticos y predictivos, como Regresión Lineal, SVM o K-Means, requieren matrices matemáticas completas y no toleran entradas vacías (`NaN`). Este paso garantiza la compatibilidad técnica del dataset con las librerías de modelado de Scikit-Learn.

Como consideración metodológica, en un flujo productivo ideal los parámetros de imputación deberían ajustarse con el conjunto de entrenamiento y luego aplicarse al conjunto de prueba. Esta observación se retoma en la Fase 5 para fortalecer la evaluación del modelo.

### 4.3 Codificación de Variables Categóricas

Los algoritmos de Machine Learning son, en su núcleo, funciones matemáticas. Por lo tanto, son incapaces de interpretar directamente cadenas de texto como "Yes", "No", o direcciones cardinales como "NNE" o "SW". Para que el modelo pueda aprender de esta información, debemos traducir estos valores categóricos a representaciones numéricas.

Aplicaremos dos estrategias de transformación según la naturaleza de la variable:

1. **Mapeo Binario:** Para variables de dos estados "`RainToday` y `RainTomorrow`", las convertiremos directamente en variables lógicas usando `0` para "No" y `1` para "Yes".
2. **One-Hot Encoding (Variables Dummy):** Para variables nominales sin un orden jerárquico natural (como `Location` y las direcciones del viento `WindGustDir`, `WindDir9am`, `WindDir3pm`), crearemos nuevas columnas binarias para cada categoría. Esto evita que el modelo asuma falsamente que un viento del Norte vale "más" que un viento del Sur.
3. **Ingeniería de Características (Feature Engineering) en la Fecha:** La columna `Date` no puede usarse cruda. Extraeremos el "Mes" para capturar la estacionalidad del clima y eliminaremos la fecha exacta, simplificando el aprendizaje del modelo.

```python
# 1. Ingeniería de Características: Extraer el Mes y eliminar 'Date'
if 'Date' in df_preparado.columns:
    df_preparado['Date'] = pd.to_datetime(df_preparado['Date'])
    df_preparado['Month'] = df_preparado['Date'].dt.month
    df_preparado = df_preparado.drop(columns=['Date'])
    print("Columna 'Date' procesada: Se extrajo el 'Month' (Mes).")

# 2. Mapeo Binario para las variables de lluvia
mapeo_binario = {'No': 0, 'Yes': 1}
if 'RainToday' in df_preparado.columns and df_preparado['RainToday'].dtype == 'object':
    df_preparado['RainToday'] = df_preparado['RainToday'].map(mapeo_binario)
if 'RainTomorrow' in df_preparado.columns and df_preparado['RainTomorrow'].dtype == 'object':
    df_preparado['RainTomorrow'] = df_preparado['RainTomorrow'].map(mapeo_binario)
print("Variables 'RainToday' y 'RainTomorrow' codificadas a 0 y 1.")

# 3. One-Hot Encoding para las variables nominales (Location y Vientos)
columnas_nominales = ['Location', 'WindGustDir', 'WindDir9am', 'WindDir3pm']

columnas_presentes = [col for col in columnas_nominales if col in df_preparado.columns]

if columnas_presentes:
    df_preparado = pd.get_dummies(df_preparado, columns=columnas_presentes, drop_first=True)
    print(f"One-Hot Encoding aplicado a: {columnas_presentes}")

print("\n--- ESTADO DEL DATASET TRAS EL ENCODING ---")
print(f"Nuevas dimensiones del dataset: {df_preparado.shape}")
print("Verificación de tipos de datos (deben ser numéricos o booleanos):")
print(df_preparado.dtypes.value_counts())
```

La transformación de las variables cualitativas a un formato numérico marca un punto de inflexión estructural en la preparación del conjunto de datos. El impacto de este proceso se resume en los siguientes puntos clave:

1. **Traducción Matemática para el Modelado:** Pasamos de tener cadenas de texto irreconocibles para los algoritmos a un dataset 100% compuesto por variables numéricas y booleanas. Esto garantiza que modelos estadísticos basados en distancias (como K-Means) o en optimización de coeficientes (como Regresión Logística y SVM) puedan asimilar la información sin arrojar errores de compatibilidad.
2. **Preservación de la Naturaleza Nominal:** Al aplicar *One-Hot Encoding* sobre variables como `Location` y las direcciones del viento, evitamos introducir un orden jerárquico artificial. Si hubiéramos asignado números secuenciales (ej. Norte=1, Sur=2, Este=3), el modelo asumiría erróneamente que el Este "vale el triple" que el Norte. Mediante la creación de variables *dummy* (binarias), preservamos el peso equitativo y la identidad pura de cada categoría original.
3. **Expansión Dimensional Justificada:** El crecimiento del dataset a 107 columnas es el comportamiento técnico esperado. Aunque la dimensionalidad aumentó significativamente, cada nueva columna booleanizada aporta información específica y sin sesgos, lo que permitirá a los modelos capturar patrones climáticos detallados (por ejemplo, si la probabilidad de lluvia se dispara cuando el viento proviene de una ciudad y dirección exactas).

### 4.4 Transformación de Variables Temporales

En el paso anterior extrajimos el mes (`Month`) a partir de la fecha original para capturar la estacionalidad del clima. Sin embargo, mantener el mes como un simple número lineal del 1 al 12 presenta un problema lógico grave para los modelos de Machine Learning.

En una escala lineal, el algoritmo interpretará matemáticamente que el mes 12 (Diciembre) está extremadamente lejos del mes 1 (Enero). Sin embargo, en la realidad meteorológica, Diciembre y Enero son meses consecutivos y comparten características climáticas muy similares (pleno verano en Australia).

Para solucionar este salto y preservar la naturaleza continua del tiempo, aplicaremos una **Transformación Cíclica**. Utilizaremos funciones trigonométricas (Seno y Coseno) para mapear los 12 meses en un círculo bidimensional. De esta manera, el modelo comprenderá perfectamente que después del mes 12, el ciclo se cierra y vuelve a comenzar en el mes 1, manteniendo la distancia matemática real entre las estaciones del año.

```python
print("--- TRANSFORMACIÓN CÍCLICA DEL MES ---")

if 'Month' in df_preparado.columns:
    df_preparado['Month_sin'] = np.sin(2 * np.pi * df_preparado['Month'] / 12)
    df_preparado['Month_cos'] = np.cos(2 * np.pi * df_preparado['Month'] / 12)

    df_preparado = df_preparado.drop(columns=['Month'])

    print("Transformación exitosa: Se crearon las variables 'Month_sin' y 'Month_cos'.")
    print("La columna lineal original 'Month' fue eliminada.\n")
else:
    print("La columna 'Month' no se encuentra en el dataset.\n")

print(f"Dimensiones actuales del dataset: {df_preparado.shape}")
print("\nEjemplo de los valores transformados (Seno y Coseno):")
display(df_preparado[['Month_sin', 'Month_cos']].head())
```

La aplicación de la transformación cíclica sobre la variable "Mes" resuelve una de las inconsistencias matemáticas más comunes en el análisis de series de tiempo y estacionalidad. El impacto de este proceso se sustenta en tres pilares analíticos:

1. **Preservación de la Continuidad Temporal:** Al reemplazar una escala lineal (1 al 12) por coordenadas trigonométricas (Seno y Coseno), logramos que el modelo entienda que Diciembre (mes 12) y Enero (mes 1) están a la misma "distancia" matemática que Enero y Febrero. Esto evita la falsa interpretación algorítmica de que el fin de año y el inicio del siguiente representan extremos climáticos opuestos.
2. **Captura Precisa de la Estacionalidad:** El clima es un fenómeno intrínsecamente cíclico. Representar el tiempo como un círculo bidimensional permite a los modelos de Machine Learning (especialmente redes neuronales y algoritmos basados en árboles) identificar de manera más natural y eficiente los patrones repetitivos de las estaciones (verano, otoño, invierno, primavera).
3. **Optimización del Aprendizaje Predictivo:** Al eliminar el sesgo lineal, evitamos que modelos estadísticos asuman relaciones falsas (por ejemplo, pensar que "a mayor número de mes, mayor temperatura" de manera indefinida). Esto mejora drásticamente la capacidad de generalización del modelo frente a datos futuros.

### 4.5 Tratamiento de Outliers

Durante la **Fase 2 Análisis Exploratorio de Datos**, identificamos mediante diagramas de caja (Boxplots) la presencia de una gran cantidad de valores atípicos, especialmente en variables como `Rainfall` (Precipitaciones) y `WindGustSpeed` (Velocidad de ráfagas de viento). Estas variables presentan una fuerte asimetría positiva debido a eventos climáticos extremos (tormentas severas).

**Estrategia de Tratamiento: Winsorización (Capping)**
Dado que estos valores no son errores de medición, sino eventos meteorológicos reales y valiosos, **eliminar esas filas sería un error grave**, ya que perderíamos la información de los días con peores condiciones climáticas (que suelen ser los que más nos interesa predecir).

En su lugar, aplicaremos una técnica de **Winsorización basada en el Rango Intercuartílico (IQR)**. Esto significa que calcularemos un límite superior (utilizando un multiplicador conservador de 3.0 para detectar solo los extremos más severos) y "recortaremos" o "taparemos" (Capping) los valores que superen ese límite, igualándolos al valor máximo permitido. De este modo, conservamos el registro del día tormentoso, pero evitamos que un valor numérico desproporcionado distorsione los coeficientes de los algoritmos de aprendizaje.

```python
variables_outliers = ['Rainfall', 'WindGustSpeed', 'Evaporation', 'WindSpeed9am', 'WindSpeed3pm']

variables_presentes = [col for col in variables_outliers if col in df_preparado.columns]

print("--- APLICANDO TRATAMIENTO DE OUTLIERS (CAPPING) ---")

for col in variables_presentes:
    Q1 = df_preparado[col].quantile(0.25)
    Q3 = df_preparado[col].quantile(0.75)
    IQR = Q3 - Q1

    limite_superior = Q3 + 3.0 * IQR

    cantidad_outliers = (df_preparado[col] > limite_superior).sum()

    df_preparado[col] = df_preparado[col].clip(upper=limite_superior)

    print(f"Variable '{col}': {cantidad_outliers} valores extremos limitados a {limite_superior:.2f}")

print("-" * 55)
print(f"Dimensiones del dataset tras el tratamiento: {df_preparado.shape} (No se perdieron filas).")
```

La aplicación de la técnica de **capping** o **winsorización** permitió controlar la influencia de valores atípicos sin eliminar registros del conjunto de datos. Esta decisión es coherente con el contexto meteorológico del caso, ya que en variables como `Rainfall`, `WindGustSpeed`, `WindSpeed9am` y `WindSpeed3pm` los valores extremos no necesariamente representan errores, sino que pueden corresponder a eventos climáticos reales, como lluvias intensas, ráfagas fuertes de viento o condiciones meteorológicas poco frecuentes.

En este proyecto se decidió aplicar capping en lugar de eliminar observaciones por las siguientes razones:

1. **Preservación de eventos climáticos relevantes.**  
   Eliminar filas completas habría reducido el tamaño del dataset y, además, podía borrar registros asociados a fenómenos extremos. En un problema climático, estos eventos son importantes porque ayudan a comprender situaciones de mayor riesgo. Por lo tanto, conservarlos de forma controlada permite mantener información valiosa sin dejar que dominen el comportamiento del modelo.

2. **Reducción de la influencia excesiva de valores extremos.**  
   Algunas variables presentan distribuciones asimétricas, especialmente `Rainfall`, donde predominan los días secos y existen pocos eventos de lluvia muy intensa. Si estos valores se mantienen sin tratamiento, pueden afectar el escalamiento y el aprendizaje de modelos sensibles a la distancia o a la magnitud de las variables. Con el capping, los valores extremos no desaparecen, pero quedan limitados a un umbral razonable.

3. **Criterio conservador mediante IQR ampliado.**  
   Se utilizó un límite superior basado en `Q3 + 3.0 * IQR`, lo que corresponde a un criterio más flexible que el umbral clásico de `1.5 * IQR`. Esta decisión evita tratar como atípicos demasiados registros válidos y permite intervenir principalmente los casos más extremos.

4. **Beneficio para el modelado y el escalamiento posterior.**  
   Al reducir la influencia de valores muy altos antes de aplicar `MinMaxScaler`, se obtiene un conjunto de datos más estable para el entrenamiento. Esto favorece la comparación entre modelos, mejora la consistencia del escalamiento y evita que unos pocos registros extremos tengan un peso desproporcionado en el análisis. Esta decisión es especialmente importante para modelos sensibles a la magnitud de las variables, como SVM, K-Means y Regresión Lineal.

5. **Consideración metodológica.**  
   En un flujo productivo completo, los umbrales de outliers deberían calcularse únicamente sobre el conjunto de entrenamiento y luego aplicarse al conjunto de prueba, para evitar fuga de información. En este cuaderno, el tratamiento se utiliza como una etapa académica de preparación general de datos. Para futuras mejoras, sería recomendable comparar este enfoque con alternativas como transformación logarítmica (`log1p`) o capping por percentiles, especialmente en variables con distribuciones muy asimétricas.

En síntesis, el capping permitió equilibrar dos necesidades: **conservar información climática relevante** y, al mismo tiempo, **reducir el impacto excesivo de valores extremos** sobre el proceso de modelado.

### 4.6 Escalamiento de Datos

Tras haber limpiado los nulos, codificado las variables categóricas, transformado las fechas y corregido los valores atípicos, nos enfrentamos a un último desafío numérico: la disparidad de escalas.

La secuencia aplicada es intencional: primero se controlaron los valores atípicos mediante capping o winsorización y luego se realizará el escalamiento. Esto es importante porque `MinMaxScaler` es sensible a los valores mínimos y máximos; si se aplicara antes de tratar los outliers, unos pocos eventos extremos podrían comprimir artificialmente la mayor parte de los datos en un rango muy pequeño. Al realizar el capping previamente, se conserva la información climática extrema de forma controlada y se obtiene una normalización más estable para los modelos.

Actualmente, nuestras variables operan en magnitudes completamente distintas. Por ejemplo, la presión atmosférica ronda valores de 1000 a 1030 hPa, la humedad varía de 0 a 100, y nuestras variables codificadas (trigonométricas y *dummies*) oscilan entre -1 y 1 o son simplemente 0 y 1.

Es absolutamente necesario **normalizar la escala de las variables numéricas** para evitar que aquellas con rangos numéricos más amplios (como la presión) dominen artificialmente el proceso de aprendizaje por sobre las demás.

**Impacto en Algoritmos Sensibles:**
El escalamiento resulta esencial para garantizar que todas las variables contribuyan de manera equilibrada al entrenamiento. Esto es particularmente crítico para algoritmos como **Support Vector Machines (SVM)** y **K-Means**. Estos modelos son altamente sensibles a la magnitud de las variables, ya que basan gran parte de su funcionamiento en el cálculo de distancias matemáticas (como la distancia euclidiana) en un espacio multidimensional. Si no escalamos los datos, una variable con valores en los miles anulará por completo la influencia de una variable binaria.

A continuación, aplicaremos una transformación (`MinMaxScaler`) que comprimirá todas las variables predictoras a un rango uniforme y estandarizado (entre 0 y 1).

```python
print("--- APLICANDO ESCALAMIENTO DE DATOS ---")

y = df_preparado['RainTomorrow']
X = df_preparado.drop(columns=['RainTomorrow'])

scaler = MinMaxScaler()

X_escalado_array = scaler.fit_transform(X)

df_escalado = pd.DataFrame(X_escalado_array, columns=X.columns, index=X.index)

df_escalado['RainTomorrow'] = y

print(f"Escalamiento exitoso. Dimensiones finales del dataset: {df_escalado.shape}")
print("\nEjemplo de los primeros registros (todas las variables unificadas entre 0 y 1):")
display(df_escalado.head())
```

La normalización aplicada mediante la técnica de escalamiento lineal (`MinMaxScaler`) ha consolidado con éxito la homogeneidad métrica del dataset. El impacto de este procedimiento en la calidad de los datos se traduce en los siguientes beneficios analíticos:

1. **Equilibrio de Influencia Proporcional:** Al transformar todas las características numéricas continuas al rango uniforme de 0 a 1, eliminamos el sesgo de magnitud. Variables con rangos originalmente extensos (como la presión barométrica en los miles o la velocidad del viento en decenas) ya no dominarán artificialmente el proceso de optimización matemática por sobre variables de rango acotado, garantizando que cada predictor aporte información de manera equitativa durante el aprendizaje.
2. **Optimización para Algoritmos Basados en Distancias:** Este paso es críticamente obligatorio para modelos basados en medidas de similitud espacial y geométrica, tales como **Support Vector Machine (SVM)** y **K-Means**. Dado que estos algoritmos calculan la cercanía entre observaciones utilizando fórmulas de distancia (como la distancia euclidiana), omitir el escalamiento causaría que la variable de mayor magnitud anulara por completo la influencia de las demás. Gracias a la normalización, los cálculos de distancia reflejan rigurosamente las relaciones de proximidad y similitud real de los patrones climáticos.
3. **Estabilidad y Convergencia Numérica:** Reducir la dispersión de las magnitudes a una escala común estabiliza el comportamiento matemático de las funciones de pérdida, acelerando la convergencia de los algoritmos de optimización (como el descenso de gradiente) y garantizando entrenamientos mucho más rápidos, estables y confiables.

### 4.7 División de Datos en Entrenamiento y Prueba (Train/Test Split)

Para culminar de forma rigurosa la **Fase 3: Preparación de los Datos**, realizaremos la partición del conjunto de datos definitivo en dos subconjuntos independientes: **Train** y **Test**.

En cualquier proyecto científico de Machine Learning, entrenar un modelo y evaluar su rendimiento utilizando exactamente los mismos datos constituye un error metodológico severo conocido como sobreajuste (*Overfitting*). Si hiciéramos eso, el modelo simplemente memorizaría los registros históricos de lluvia, obteniendo métricas excelentes en desarrollo, pero fallando por completo al intentar predecir en el mundo real.

**Objetivo de la Partición:**
* **Subconjunto de Entrenamiento (80% de los datos):** Será el entorno controlado expuesto a los algoritmos de minería de datos para que descubran las reglas, coeficientes y patrones climáticos precursores de las precipitaciones.
* **Subconjunto de Prueba (20% de los datos):** Actuará como un examen a ciegas. Contiene registros que el modelo jamás verá durante su entrenamiento, sirviendo como el estándar dorado para evaluar de manera objetiva e imparcial su verdadera **capacidad de generalización** ante información completamente nueva.

Para asegurar la reproducibilidad de este experimento, fijaremos una semilla aleatoria (`random_state=42`) y aplicaremos una **estratificación** basada en nuestra variable objetivo (`RainTomorrow`), garantizando que la proporción de días lluviosos y secos se mantenga idéntica en ambas porciones.

```python
print("--- EJECUTANDO PARTICIÓN DE DATOS ---")

X_final = df_escalado.drop(columns=['RainTomorrow'])
y_final = df_escalado['RainTomorrow']

X_train, X_test, y_train, y_test = train_test_split(
    X_final, y_final,
    test_size=0.20,
    random_state=42,
    stratify=y_final
)

print(f"Dataset Total       : {X_final.shape[0]} registros, {X_final.shape[1]} características.")
print(f"Subconjunto Entreno (X_train): {X_train.shape[0]} registros")
print(f"Subconjunto Prueba  (X_test) : {X_test.shape[0]} registros\n")

print("Distribución porcentual de la variable objetivo (Estratificación):")
print(f"Proporción original en el dataset : \n{y_final.value_counts(normalize=True) * 100}\n")
print(f"Proporción en el set de Entrenamiento : \n{y_train.value_counts(normalize=True) * 100}\n")
print(f"Proporción en el set de Prueba        : \n{y_test.value_counts(normalize=True) * 100}\n")

print("¡Fase 3 concluida exitosamente! Los datos están listos para la Fase 4: Modelado.")
```

La partición del conjunto de datos se ha ejecutado exitosamente, separando un 80% de los registros para la fase de aprendizaje y reservando un 20% intacto para la fase de validación. Esta división es un pilar fundamental para garantizar la validez científica del proyecto por las siguientes razones:

1. **Evaluación Objetiva de la Generalización:** El objetivo del Machine Learning no es memorizar el pasado, sino predecir el futuro. Al aislar un conjunto de prueba (*Test*), obligamos al modelo a enfrentarse a escenarios climáticos que jamás ha visto. Esto nos permite medir de manera realista y objetiva qué tan bien generalizará sus predicciones cuando sea puesto en producción con datos de nuevos días.
2. **Prevención del Sobreajuste (Overfitting):** Si evaluáramos el modelo con los mismos datos con los que fue entrenado, obtendríamos métricas de rendimiento falsamente altas. La partición reduce este riesgo metodológico y permite comparar desempeño en entrenamiento y prueba.
3. **Estratificación:** Tal como demuestran los resultados, la proporción de la clase minoritaria (días de lluvia, ~22,4%) se ha mantenido de forma similar en ambos subconjuntos. Esto asegura que el algoritmo aprenderá de una muestra representativa y será evaluado bajo las mismas condiciones de desbalance natural del clima australiano.

**Nota metodológica de buena práctica:** en un flujo productivo ideal, los procesos de imputación, tratamiento de outliers y escalamiento deben ajustarse únicamente con el conjunto de entrenamiento y luego aplicarse al conjunto de prueba. En este notebook se mantiene la estructura original de la preparación para conservar la continuidad del trabajo desarrollado en la Evaluación 2; sin embargo, la Fase 5 incorpora esta consideración como una limitación metodológica relevante, ya que evita interpretar las métricas como resultados definitivos de producción.

**Conclusión de la Fase 3:**
Con esta etapa, se da por finalizada formalmente la **Fase de Preparación de los Datos**. El dataset original ha sido sometido a un riguroso proceso de limpieza de valores nulos, codificación de variables categóricas, transformación cíclica temporal, tratamiento de outliers y escalamiento numérico. Los datos se encuentran ahora en un formato matemático apto para ser utilizados por los algoritmos predictivos y de segmentación.

### 5.1 Modelo de Regresión Lineal para Predicción de RISK_MM

El primer modelo que abordaremos es la **Regresión Lineal Múltiple**. Este es el algoritmo estadístico estándar para predecir variables numéricas continuas. Nuestro objetivo aquí no es clasificar, sino responder a la pregunta: *¿Cuántos milímetros de agua caerán mañana?*

**Consideración Metodológica:**
Durante la Fase 3, eliminamos la variable `RISK_MM` del dataset de entrenamiento principal para evitar fugas de información. Para este modelo específico, rescataremos `RISK_MM` del dataset original, emparejándola con los mismos registros que ya escalamos y preparamos en nuestros conjuntos `X_train` y `X_test`. Evaluaremos el modelo utilizando el Error Cuadrático Medio (MSE) y el Coeficiente de Determinación ($R^2$).

```python
# 1. ENTRENAMIENTO - REGRESIÓN LINEAL (TARGET: RISK_MM)
print("--- 1. PREPARANDO DATOS Y ENTRENANDO EL MODELO ---")

y_train_reg = df.loc[X_train.index, 'RISK_MM']
y_test_reg = df.loc[X_test.index, 'RISK_MM']

y_train_reg = y_train_reg.fillna(0)
y_test_reg = y_test_reg.fillna(0)

modelo_reg_lineal = LinearRegression()

modelo_reg_lineal.fit(X_train, y_train_reg)

print("¡Modelo de Regresión Lineal entrenado exitosamente!")
```

```python
# 2. PREDICCIONES - REGRESIÓN LINEAL

print("--- 2. GENERANDO PREDICCIONES ---")

y_pred_reg = modelo_reg_lineal.predict(X_test)

y_pred_reg = [max(0, pred) for pred in y_pred_reg]

print("Predicciones generadas y ajustadas correctamente.")
```

```python
# 3. MÉTRICAS Y EVALUACIÓN - REGRESIÓN LINEAL

print("--- 3. RESULTADOS Y MÉTRICAS DE EVALUACIÓN ---")

mse = mean_squared_error(y_test_reg, y_pred_reg)
mae = mean_absolute_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print(f"Error Absoluto Medio (MAE) : {mae:.4f} mm")
print(f"Error Cuadrático Medio (MSE): {mse:.4f}")
print(f"Coeficiente de Determinación (R^2): {r2:.4f}\n")

plt.figure(figsize=(10, 5))
plt.scatter(range(100), y_test_reg.iloc[:100], color='blue', label='Lluvia Real (RISK_MM)', alpha=0.6)
plt.plot(range(100), y_pred_reg[:100], color='red', label='Predicción del Modelo', linewidth=2)

plt.title('Regresión Lineal: Lluvia Real vs Predicción (Muestra de 100 días)', fontsize=14)
plt.ylabel('Cantidad de Lluvia (mm)', fontsize=12)
plt.xlabel('Observaciones (Días)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

El modelo de Regresión Lineal fue entrenado para predecir la cantidad de lluvia en milímetros (`RISK_MM`). A partir de los resultados obtenidos, se observa que este modelo funciona principalmente como una primera línea base, pero no logra capturar con alta precisión la complejidad del fenómeno climático.

#### 1. Análisis de las métricas de evaluación

**MAE (Error Absoluto Medio - 2.9888 mm):**

El modelo presenta un error promedio de aproximadamente **2.99 mm** respecto al valor real de lluvia. Esta métrica es la más fácil de interpretar, ya que indica cuánto se equivoca el modelo, en promedio, al estimar la cantidad de precipitación. Aunque el error no es extremadamente alto para días con lluvia ligera, sigue siendo una diferencia relevante cuando se busca estimar con precisión el volumen de agua registrado.

**MSE (Error Cuadrático Medio - 51.2511):**

El MSE penaliza con mayor fuerza los errores grandes. El valor obtenido indica que el modelo tiene dificultades importantes cuando aparecen eventos de lluvia más intensa, ya que estos casos generan desviaciones mayores entre el valor real y el valor predicho. Esto confirma que la Regresión Lineal no se adapta bien a comportamientos extremos o no lineales.

**Coeficiente de Determinación ($R^2$ - 0.2058):**

El valor de $R^2$ indica que el modelo explica aproximadamente el **20.58% de la variabilidad** de `RISK_MM`. En otras palabras, la mayor parte de la variación en la cantidad de lluvia no está siendo explicada por una relación lineal simple entre las variables meteorológicas utilizadas. Esto no significa que el modelo sea inútil, pero sí evidencia que su capacidad predictiva es limitada.

#### 2. Evaluación de la capacidad predictiva

**Fortalezas:**

La Regresión Lineal permite establecer una referencia inicial para el problema de predicción numérica. Su principal valor dentro del proyecto es que entrega una primera aproximación cuantitativa y facilita comprobar si las variables preparadas poseen alguna relación lineal con la cantidad de lluvia.

**Limitaciones:**

La lluvia es un fenómeno altamente variable y no lineal. Factores como humedad, presión, viento, ubicación y condiciones atmosféricas pueden interactuar de manera compleja. Por esta razón, una función lineal no logra representar adecuadamente los picos de lluvia ni los cambios bruscos propios de eventos climáticos más intensos.

#### 3. Conclusión sobre el desempeño

Los resultados **no son suficientemente sólidos para seleccionar la Regresión Lineal como modelo principal**. Su $R^2$ de **0.2058** muestra que el modelo explica una proporción baja de la variabilidad de la lluvia registrada. Sin embargo, su aplicación sigue siendo útil como modelo base, ya que permite justificar el avance hacia modelos más flexibles y no lineales, como Árboles de Decisión, Random Forest o modelos de ensamble, que pueden capturar mejor las interacciones entre variables climáticas.

### 5.2 Modelo Naive Bayes para Clasificación de RainTomorrow

El segundo modelo que implementaremos es **Naive Bayes**, un algoritmo de clasificación probabilístico fundamentado en el Teorema de Bayes. Su objetivo será predecir si lloverá o no al día siguiente (`RainTomorrow` = 1 o 0).

**¿Cómo funciona y por qué "Naive"?**
El modelo calcula la probabilidad de que ocurra un evento (lluvia) basándose en el conocimiento previo de las condiciones relacionadas (las variables meteorológicas). Se le denomina "ingenuo" porque asume una **independencia condicional absoluta** entre todas las variables predictoras. Es decir, matemáticamente asume que la humedad no tiene relación con la presión atmosférica, lo cual sabemos que en la meteorología del mundo real es falso.

A pesar de esta suposición aparentemente irreal, Naive Bayes suele ofrecer resultados sólidos como modelo base. Sus mayores fortalezas radican en que es un algoritmo extremadamente rápido, requiere bajos recursos computacionales y maneja con gran eficacia conjuntos de datos de alta dimensionalidad (como nuestro dataset actual, que posee más de 100 características tras aplicar *One-Hot Encoding*).

Para este caso, utilizaremos la variante **Gaussian Naive Bayes**, que es la más adecuada cuando se trabaja con características que han sido tratadas y estandarizadas de forma continua, asumiendo que los valores de cada clase siguen una distribución gaussiana.

```python
# 1. ENTRENAMIENTO Y PREDICCIONES

print("--- 1. ENTRENANDO EL MODELO NAIVE BAYES ---")

modelo_nb = GaussianNB()

modelo_nb.fit(X_train, y_train)

y_pred_nb = modelo_nb.predict(X_test)

print("¡Modelo Naive Bayes entrenado y predicciones generadas exitosamente!")
```

```python
# 2. EVALUACIÓN Y MÉTRICAS
print("--- 2. RESULTADOS Y MÉTRICAS DE EVALUACIÓN ---")

accuracy_nb = accuracy_score(y_test, y_pred_nb)
print(f"Exactitud (Accuracy) del Modelo: {accuracy_nb * 100:.2f}%\n")

print("Reporte de Clasificación (Precision, Recall, F1-Score):")
print(classification_report(y_test, y_pred_nb, target_names=['No Llueve (0)', 'Sí Llueve (1)']))

plt.figure(figsize=(7, 5))
cm_nb = confusion_matrix(y_test, y_pred_nb)
sns.heatmap(cm_nb, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['Predicción: No', 'Predicción: Sí'],
            yticklabels=['Realidad: No', 'Realidad: Sí'])

plt.title('Matriz de Confusión - Modelo Naive Bayes', fontsize=14)
plt.ylabel('Valor Real', fontsize=12)
plt.xlabel('Predicción del Modelo', fontsize=12)
plt.tight_layout()
plt.show()
```

Si este modelo se utilizara en un contexto agrícola donde es preferible prepararse para una lluvia que no ocurre, en lugar de ser sorprendidos por una tormenta imprevista, este comportamiento sería muy valioso.

Limitaciones: La suposición "ingenua" (naive) del algoritmo penaliza severamente el desempeño en este caso. Al asumir que todas las variables (como la humedad, la nubosidad y la presión) son independientes entre sí, el modelo sobreestima la probabilidad de lluvia cuando detecta un solo factor levemente alterado (por ejemplo, alta humedad), generando esa masiva cantidad de Falsos Positivos (9,196). En la realidad, la meteorología depende de la interacción conjunta de estas variables.

**¿Es satisfactorio el desempeño?**

Desde el punto de vista metodológico, sí, ya que nos sirve como una excelente línea base (baseline) probabilística y demuestra de forma práctica por qué la independencia de variables no aplica en el clima. Desde el punto de vista operativo, no es suficiente, ya que su baja precisión general limita su confiabilidad en el mundo real, debido a las constantes falsas alarmas.

Estos resultados justifican plenamente la necesidad de avanzar hacia algoritmos más complejos, como los Árboles de Decisión o las Máquinas de Vectores de Soporte (SVM), que sí son capaces de aprender las relaciones no lineales y dependientes entre las variables climáticas.

### 5.3 Árbol de Decisión para la Clasificación de RainTomorrow

Tras observar que el modelo Naive Bayes presentaba limitaciones debido a su suposición de independencia entre variables, avanzamos hacia un modelo de mayor complejidad estructural: el **Árbol de Decisión (Decision Tree)**.

**¿Cómo funciona?**

Un Árbol de Decisión funciona realizando particiones recursivas sobre el conjunto de datos. En cada paso o nodo, el algoritmo evalúa todas las variables y selecciona aquella que mejor separe los días lluviosos de los días secos, utilizando métricas matemáticas como la *Impureza de Gini* o la *Ganancia de Información*. Este proceso de realizar "preguntas" (ej. ¿Es la humedad a las 3pm mayor al 60%?) se repite hasta llegar a una hoja final que emite la predicción (1 = Sí llueve, 0 = No llueve).

**Pertinencia para nuestro caso de estudio:**
1. **Captura de Relaciones No Lineales:** A diferencia de la Regresión Logística o Lineal, los árboles de decisión no asumen que la relación entre la variable predictora y el objetivo es una línea recta.
2. **Interacción entre Variables:** El árbol entiende perfectamente que un día con alta presión atmosférica *combinado* con baja humedad tiene un pronóstico distinto que si se analizan ambas variables por separado (superando la debilidad de Naive Bayes).
3. **Manejo del Desbalance:** Para este entrenamiento, configuraremos el hiperparámetro `class_weight='balanced'`, lo cual obligará al algoritmo a prestarle más atención matemática a nuestra clase minoritaria (los días de lluvia). Además, limitaremos su profundidad (`max_depth=10`) para evitar que el modelo memorice los datos y caiga en sobreajuste (*Overfitting*).

```python
# 1. ENTRENAMIENTO - ÁRBOL DE DECISIÓN

print("--- 1. ENTRENANDO EL MODELO DE ÁRBOL DE DECISIÓN ---")

modelo_arbol = DecisionTreeClassifier(
    max_depth=10,
    class_weight='balanced',
    random_state=42
)

# 2. Entrenar el modelo con el subconjunto de entrenamiento
modelo_arbol.fit(X_train, y_train)

print("¡Modelo de Árbol de Decisión entrenado exitosamente!")
```

```python
# 2. EVALUACIÓN Y MÉTRICAS - ÁRBOL DE DECISIÓN

print("--- 2. RESULTADOS Y MÉTRICAS DE EVALUACIÓN ---")

y_pred_arbol = modelo_arbol.predict(X_test)

accuracy_arbol = accuracy_score(y_test, y_pred_arbol)
print(f"Exactitud (Accuracy) del Modelo: {accuracy_arbol * 100:.2f}%\n")

print("Reporte de Clasificación detallado:")
print(classification_report(y_test, y_pred_arbol, target_names=['No Llueve (0)', 'Sí Llueve (1)']))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

cm_arbol = confusion_matrix(y_test, y_pred_arbol)
sns.heatmap(cm_arbol, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Predicción: No', 'Predicción: Sí'],
            yticklabels=['Realidad: No', 'Realidad: Sí'], ax=axes[0])
axes[0].set_title('Matriz de Confusión - Árbol de Decisión', fontsize=14)
axes[0].set_ylabel('Valor Real')
axes[0].set_xlabel('Predicción del Modelo')

importancias = modelo_arbol.feature_importances_
df_importancias = pd.DataFrame({'Variable': X_train.columns, 'Importancia': importancias})
df_importancias = df_importancias.sort_values(by='Importancia', ascending=False).head(10)

sns.barplot(x='Importancia', y='Variable', data=df_importancias, ax=axes[1], palette='viridis')
axes[1].set_title('Top 10 Variables más Importantes', fontsize=14)
axes[1].set_xlabel('Nivel de Importancia (Gini)')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

### 5.4 Máquinas de Vectores de Soporte (SVM)

Avanzamos en la complejidad de nuestros algoritmos implementando un modelo de **Máquinas de Vectores de Soporte o SVM**. Este es uno de los algoritmos supervisados más robustos y potentes tanto para regresión como para clasificación.

**¿Cómo funciona?**

El objetivo de SVM es encontrar un hiperplano (una frontera geométrica de decisión) en un espacio multidimensional que separe de la mejor manera posible las dos clases (Llueve / No Llueve). Para lograrlo, el algoritmo busca maximizar el "margen" o la distancia entre este hiperplano y los puntos de datos más cercanos de cada clase, los cuales son llamados **Vectores de Soporte**.

**Pertinencia para nuestro caso de estudio:**
1. **Sensibilidad a la Escala:** Como se mencionó en la Fase 3, SVM basa su funcionamiento en el cálculo de distancias matemáticas entre los puntos de datos. Si no hubiéramos escalado nuestras variables (`MinMaxScaler`), este modelo fracasaría estrepitosamente. Gracias a nuestra preparación previa, SVM está en condiciones óptimas para operar.
2. **Fronteras No Lineales (Kernel Trick):** Utilizaremos un *Kernel* de tipo RBF (Función de Base Radial), que permite al algoritmo proyectar los datos a dimensiones superiores para trazar fronteras de decisión curvas y complejas, adaptándose muy bien a la naturaleza no lineal del clima.
3. **Consideración Computacional:** SVM es un algoritmo altamente demandante en términos de CPU y memoria, especialmente con datasets masivos (más de 140,000 registros como el nuestro). Por ello, limitaremos el número máximo de iteraciones matemáticas para asegurar que el entrenamiento concluya en un tiempo razonable.

```python
# 1. ENTRENAMIENTO - SUPPORT VECTOR MACHINE (SVM)

print("--- 1. ENTRENANDO EL MODELO SVM ---")
print("Nota: SVM es computacionalmente intensivo. Se utiliza max_iter=3000 para controlar el tiempo de ejecución.")

inicio = time.time()

modelo_svm = SVC(
    kernel='rbf',
    class_weight='balanced',
    max_iter=3000,
    random_state=42
)

with warnings.catch_warnings(record=True) as advertencias_svm:
    warnings.simplefilter("always", ConvergenceWarning)
    modelo_svm.fit(X_train, y_train)

fin = time.time()

advertencias_convergencia_svm = [str(w.message) for w in advertencias_svm]

print(f"¡Modelo SVM entrenado en {fin - inicio:.2f} segundos!")
if advertencias_convergencia_svm:
    print("Advertencia metodológica: el modelo SVM generó advertencias de convergencia.")
    print("Esto significa que su desempeño debe interpretarse como una configuración experimental limitada y no como el máximo potencial del algoritmo.")
else:
    print("No se registraron advertencias de convergencia durante el entrenamiento de SVM.")
```

```python
# 2. EVALUACIÓN Y MÉTRICAS - SVM

print("--- 2. RESULTADOS Y MÉTRICAS DE EVALUACIÓN ---")

y_pred_svm = modelo_svm.predict(X_test)

accuracy_svm = accuracy_score(y_test, y_pred_svm)
print(f"Exactitud (Accuracy) del Modelo: {accuracy_svm * 100:.2f}%\n")

print("Reporte de Clasificación detallado:")
print(classification_report(y_test, y_pred_svm, target_names=['No Llueve (0)', 'Sí Llueve (1)']))

plt.figure(figsize=(7, 5))
cm_svm = confusion_matrix(y_test, y_pred_svm)
sns.heatmap(cm_svm, annot=True, fmt='d', cmap='Purples',
            xticklabels=['Predicción: No', 'Predicción: Sí'],
            yticklabels=['Realidad: No', 'Realidad: Sí'])

plt.title('Matriz de Confusión - Máquinas de Vectores de Soporte (SVM)', fontsize=14)
plt.ylabel('Valor Real (Observado)', fontsize=12)
plt.xlabel('Predicción del Modelo', fontsize=12)
plt.tight_layout()
plt.show()
```

Al evaluar las métricas generadas por el modelo de Máquinas de Vectores de Soporte (SVM), observamos un comportamiento anómalo que contrasta fuertemente con los resultados obtenidos por el Árbol de Decisión. En esta configuración experimental, el modelo no logró una frontera de decisión útil para separar correctamente los días con lluvia y sin lluvia.

#### 1. Análisis de desempeño general
La exactitud del modelo fue considerablemente baja en comparación con los otros clasificadores. Además, al revisar el reporte de clasificación, se observa que el modelo tiende a concentrar sus predicciones en la clase positiva, lo que genera un recall muy alto para los días lluviosos, pero una precisión muy baja.

Este resultado demuestra que un recall elevado no siempre implica un buen modelo. Si el modelo clasifica casi todos los registros como lluvia, detectará muchos días lluviosos reales, pero también producirá una gran cantidad de falsas alarmas.

#### 2. Diagnóstico de las posibles causas
Este comportamiento puede estar relacionado con tres factores técnicos:

1. **Límite de iteraciones (`max_iter=3000`):** SVM es costoso computacionalmente en conjuntos de datos grandes y con muchas variables. Al limitar las iteraciones, el modelo puede detenerse antes de alcanzar una convergencia adecuada.
2. **Uso de `class_weight='balanced'`:** esta configuración busca compensar el desbalance de clases, pero en un modelo que no converge correctamente puede provocar una sobrecompensación hacia la clase minoritaria.
3. **Falta de ajuste de hiperparámetros:** los parámetros por defecto del kernel RBF no necesariamente son los más adecuados para este conjunto de datos.

### 5.5 Modelo K-Means Clustering

Para finalizar la fase de modelado, implementaremos **K-Means**, el algoritmo de agrupamiento (clustering) más popular en el aprendizaje no supervisado.

**¿En qué consiste el proceso de Clustering?**
A diferencia de los modelos supervisados anteriores, donde predecíamos una etiqueta conocida como 0 o 1, el clustering busca descubrir estructuras ocultas en los datos sin utilizar una variable objetivo. El algoritmo agrupa las observaciones basándose en su similitud matemática.

**¿Cómo funciona K-Means?**
1. **Inicialización:** Se seleccionan al azar $K$ puntos centrales llamados centroides.
2. **Asignación:** Cada registro meteorológico se asigna al centroide más cercano, calculando la distancia espacial entre sus variables.
3. **Actualización:** Los centroides se recalculan moviéndose al centro de los puntos que les fueron asignados.
4. **Iteración:** Los pasos 2 y 3 se repiten hasta que los centroides dejan de moverse.

**Consideraciones metodológicas:**
* **Escalamiento:** K-Means es sensible a las magnitudes de los datos porque utiliza distancias geométricas. Por ello, el uso de `MinMaxScaler` es necesario para que variables como presión, humedad, temperatura o viento sean comparables.
* **Selección de variables:** Para mejorar la interpretación de los clústeres, se priorizan variables climáticas principales y se evita que el agrupamiento dependa excesivamente de variables categóricas codificadas con One-Hot Encoding.
* **Método del Codo:** Dado que el algoritmo no sabe cuántos grupos formar, se ejecuta una prueba variando el número de clústeres ($K$) del 1 al 10. El objetivo es identificar el punto donde agregar más clústeres deja de aportar mejoras relevantes en la inercia.

```python
# 1. MÉTODO DEL CODO (ELBOW METHOD) PARA K-MEANS

print("--- 1. CALCULANDO EL NÚMERO ÓPTIMO DE CLÚSTERES ---")

# Para K-Means se priorizan variables climáticas principales.
# Esto evita que el agrupamiento quede dominado por variables categóricas codificadas mediante One-Hot Encoding.
variables_kmeans_base = [
    'MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed',
    'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 'Humidity3pm',
    'Pressure3pm', 'RainToday', 'Month_sin', 'Month_cos'
]

variables_kmeans = [col for col in variables_kmeans_base if col in X_train.columns]
X_kmeans_train = X_train[variables_kmeans].copy()

print("Variables utilizadas para K-Means:")
print(variables_kmeans)

inercia = []
rango_k = range(1, 11)

for k in rango_k:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_temp.fit(X_kmeans_train)
    inercia.append(kmeans_temp.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(rango_k, inercia, marker='o', linestyle='--', linewidth=2, markersize=8)
plt.title('Método del Codo para determinar K óptimo', fontsize=14)
plt.xlabel('Número de Clústeres (K)', fontsize=12)
plt.ylabel('Inercia (Suma de errores cuadráticos)', fontsize=12)
plt.xticks(rango_k)
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 2. ENTRENAMIENTO K-MEANS, VISUALIZACIÓN Y PERFIL DE CLÚSTERES

print("--- 2. ENTRENANDO K-MEANS Y GENERANDO CLÚSTERES ---")

k_optimo = 3

kmeans_final = KMeans(n_clusters=k_optimo, random_state=42, n_init=10)
clusters_generados = kmeans_final.fit_predict(X_kmeans_train)

# Visualización usando dos variables meteorológicas interpretables.
df_visual = pd.DataFrame({
    'Humedad a las 3pm': X_kmeans_train['Humidity3pm'],
    'Presión a las 3pm': X_kmeans_train['Pressure3pm'],
    'Cluster': clusters_generados
})

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_visual,
    x='Humedad a las 3pm',
    y='Presión a las 3pm',
    hue='Cluster',
    palette='Set1',
    alpha=0.6,
    s=30
)

centroides_humedad = kmeans_final.cluster_centers_[:, X_kmeans_train.columns.get_loc('Humidity3pm')]
centroides_presion = kmeans_final.cluster_centers_[:, X_kmeans_train.columns.get_loc('Pressure3pm')]

plt.scatter(
    centroides_humedad,
    centroides_presion,
    color='black',
    marker='X',
    s=200,
    label='Centroides'
)

plt.title(f'Visualización de Clústeres (K={k_optimo}): Humedad vs Presión', fontsize=14)
plt.xlabel('Humedad a las 3pm (escalada)', fontsize=12)
plt.ylabel('Presión a las 3pm (escalada)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Distribución de registros por clúster:")
distribucion_clusters = pd.Series(clusters_generados).value_counts().sort_index()
distribucion_clusters_pct = (distribucion_clusters / len(clusters_generados) * 100).round(2)

df_distribucion_clusters = pd.DataFrame({
    'Cantidad_registros': distribucion_clusters,
    'Porcentaje': distribucion_clusters_pct.astype(str) + '%'
})

display(df_distribucion_clusters)

# Perfil descriptivo de clústeres usando variables originales interpretables.
variables_perfil_modelado = [
    'Humidity3pm', 'Pressure3pm', 'Rainfall', 'WindGustSpeed', 'RainTomorrow'
]
variables_perfil_modelado = [col for col in variables_perfil_modelado if col in df_preparado.columns]

df_perfil_modelado = df_preparado.loc[X_train.index, variables_perfil_modelado].copy()
df_perfil_modelado['Cluster'] = clusters_generados

perfil_clusters = df_perfil_modelado.groupby('Cluster').agg(
    Cantidad_registros=('Cluster', 'count'),
    Humidity3pm_promedio=('Humidity3pm', 'mean'),
    Pressure3pm_promedio=('Pressure3pm', 'mean'),
    Rainfall_promedio=('Rainfall', 'mean'),
    WindGustSpeed_promedio=('WindGustSpeed', 'mean'),
    Proporcion_lluvia_manana=('RainTomorrow', 'mean')
).round(4)

perfil_clusters['Porcentaje_lluvia_manana'] = (
    perfil_clusters['Proporcion_lluvia_manana'] * 100
).round(2).astype(str) + '%'

perfil_clusters = perfil_clusters.drop(columns=['Proporcion_lluvia_manana'])

display(perfil_clusters)
```

### **6.2 Evaluación del modelo no supervisado: K-Means**

Los modelos de agrupamiento requieren una evaluación distinta a la de los modelos supervisados. En K-Means no existe una etiqueta objetivo que indique previamente a qué clúster debe pertenecer cada observación. Por lo tanto, la evaluación se concentra en dos elementos principales:

1. **Determinar un número adecuado de clústeres.**
2. **Medir la calidad del agrupamiento obtenido.**

Para determinar el número de clústeres se utiliza el **Método del Codo**, el cual analiza cómo disminuye la inercia o suma de distancias internas al aumentar el número de clústeres. La lógica es que, al aumentar K, los grupos se vuelven más pequeños y homogéneos, pero llega un punto donde agregar más clústeres aporta una mejora cada vez menor. Ese punto de cambio se interpreta como una alternativa razonable para seleccionar K.

#### 6.2.1 Método del Codo y Varianza Explicativa para K-Means

```python
print("--- MÉTODO DEL CODO Y VARIANZA EXPLICATIVA ---")

variables_kmeans_base = [
    'MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed',
    'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 'Humidity3pm',
    'Pressure3pm', 'RainToday', 'Month_sin', 'Month_cos'
]
variables_kmeans = [col for col in variables_kmeans_base if col in X_train.columns]
X_kmeans_eval = X_train[variables_kmeans].copy()

print("Variables utilizadas en la evaluación de K-Means:")
print(variables_kmeans)

rango_k = range(1, 11)
inercias = []

for k in rango_k:
    modelo_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    modelo_temp.fit(X_kmeans_eval)
    inercias.append(modelo_temp.inertia_)

inercia_k1 = inercias[0]

df_codo = pd.DataFrame({
    "K": list(rango_k),
    "Inercia_WCSS": inercias
})

df_codo["Varianza_Explicada_Aprox"] = 1 - (df_codo["Inercia_WCSS"] / inercia_k1)

df_codo["Mejora_Incremental"] = df_codo["Varianza_Explicada_Aprox"].diff()

df_codo["Varianza_Explicada_Aprox_%"] = (df_codo["Varianza_Explicada_Aprox"] * 100).round(2)
df_codo["Mejora_Incremental_%"] = (df_codo["Mejora_Incremental"] * 100).round(2)

display(df_codo)

plt.figure(figsize=(8, 5))
plt.plot(df_codo["K"], df_codo["Inercia_WCSS"], marker="o")
plt.title("Método del Codo para K-Means")
plt.xlabel("Número de clústeres K")
plt.ylabel("Inercia / WCSS")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(df_codo["K"], df_codo["Varianza_Explicada_Aprox_%"], marker="o")
plt.title("Varianza Explicativa Aproximada por número de clústeres")
plt.xlabel("Número de clústeres K")
plt.ylabel("Varianza explicada aproximada (%)")
plt.grid(True, alpha=0.3)
plt.show()
```

**Interpretación del Método del Codo y Varianza Explicativa**

El Método del Codo permite observar cómo cambia la inercia al aumentar el número de clústeres. En los resultados obtenidos, el paso de **K = 1** a **K = 2** genera la mayor mejora, con una varianza explicativa aproximada de **41,82%**. Al aumentar a **K = 3**, la varianza explicativa aproximada sube a **53,99%**, agregando una mejora incremental de **12,17 puntos porcentuales**.

A partir de **K = 4**, la mejora continúa, pero comienza a disminuir: **7,40 puntos porcentuales** entre K = 3 y K = 4, y cerca de **3,45 puntos porcentuales** entre K = 4 y K = 5. Esto indica que agregar más clústeres sigue reduciendo la inercia, pero con ganancias progresivamente menores.

En esta evaluación se mantiene **K = 3** como alternativa de segmentación porque representa un equilibrio adecuado entre reducción de variabilidad interna, simplicidad del modelo e interpretabilidad de los perfiles climáticos. La selección no se plantea como un valor matemáticamente único o definitivo, sino como una decisión metodológica razonable para construir segmentos comprensibles y útiles dentro del análisis climático.

Además, K-Means se evalúa sobre variables climáticas principales y no sobre todas las variables codificadas mediante One-Hot Encoding. Esto fortalece la coherencia del agrupamiento, ya que los clústeres quedan explicados principalmente por condiciones meteorológicas como temperatura, humedad, presión, viento, lluvia observada y estacionalidad.

#### 6.4.4 Curvas ROC y AUC

```python
def obtener_score_modelo(modelo, X):
    if hasattr(modelo, "predict_proba"):
        return modelo.predict_proba(X)[:, 1]
    elif hasattr(modelo, "decision_function"):
        return modelo.decision_function(X)
    else:
        return None

resultados_auc = []

plt.figure(figsize=(8, 6))

for nombre, modelo in modelos_clasificacion.items():
    y_score = obtener_score_modelo(modelo, X_test)

    if y_score is not None:
        fpr, tpr, thresholds = roc_curve(y_test, y_score)
        auc_modelo = auc(fpr, tpr)

        resultados_auc.append({
            "Modelo": nombre,
            "ROC_AUC": auc_modelo
        })

        plt.plot(fpr, tpr, label=f"{nombre} (AUC = {auc_modelo:.3f})")
    else:
        resultados_auc.append({
            "Modelo": nombre,
            "ROC_AUC": np.nan
        })

plt.plot([0, 1], [0, 1], linestyle="--", label="Azar")
plt.title("Curvas ROC - Modelos de Clasificación")
plt.xlabel("Tasa de Falsos Positivos")
plt.ylabel("Tasa de Verdaderos Positivos")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

df_auc = pd.DataFrame(resultados_auc)
display(df_auc)
```

#### 6.4.5 Comparación entre entrenamiento y prueba para detectar overfitting o underfitting

```python
comparacion_train_test = []

for nombre, modelo in modelos_clasificacion.items():
    y_pred_train = modelo.predict(X_train)
    y_pred_test = modelo.predict(X_test)

    comparacion_train_test.append({
        "Modelo": nombre,
        "Accuracy_Train": accuracy_score(y_train, y_pred_train),
        "Accuracy_Test": accuracy_score(y_test, y_pred_test),
        "Diferencia_Accuracy": accuracy_score(y_train, y_pred_train) - accuracy_score(y_test, y_pred_test),
        "F1_Clase_1_Train": f1_score(y_train, y_pred_train, pos_label=1, zero_division=0),
        "F1_Clase_1_Test": f1_score(y_test, y_pred_test, pos_label=1, zero_division=0),
        "Diferencia_F1": f1_score(y_train, y_pred_train, pos_label=1, zero_division=0) - f1_score(y_test, y_pred_test, pos_label=1, zero_division=0)
    })

df_train_test = pd.DataFrame(comparacion_train_test).round(4)
display(df_train_test)
```

#### 6.4.6 Validación cruzada de modelos de clasificación

```python
print("--- VALIDACIÓN CRUZADA DE MODELOS DE CLASIFICACIÓN ---")
print("Nota: la validación cruzada de SVM puede tardar varios minutos dependiendo del equipo.")

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

modelos_cv = {
    "Modelo Base": DummyClassifier(strategy="most_frequent", random_state=42),
    "Naive Bayes": GaussianNB(),
    "Árbol de Decisión": DecisionTreeClassifier(
        max_depth=10,
        class_weight="balanced",
        random_state=42
    ),
    "SVM": SVC(
        kernel="rbf",
        class_weight="balanced",
        max_iter=3000,
        random_state=42
    )
}

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

resultados_cv = []

for nombre, modelo in modelos_cv.items():
    scores = cross_validate(
        modelo,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        error_score=np.nan
    )

    resultados_cv.append({
        "Modelo": nombre,
        "Accuracy_CV_Media": np.nanmean(scores["test_accuracy"]),
        "Precision_CV_Media": np.nanmean(scores["test_precision"]),
        "Recall_CV_Media": np.nanmean(scores["test_recall"]),
        "F1_CV_Media": np.nanmean(scores["test_f1"]),
        "ROC_AUC_CV_Media": np.nanmean(scores["test_roc_auc"])
    })

df_resultados_cv = pd.DataFrame(resultados_cv).round(4)
display(df_resultados_cv)
```

   El clúster 0 se caracteriza por mayor humedad, mayor lluvia promedio y mayor proporción de lluvia al día siguiente. Por ello, se recomienda utilizar la segmentación como una herramienta complementaria para identificar registros o condiciones similares a este perfil de mayor riesgo. Esta acción no reemplaza al modelo supervisado, pero permite comprender mejor los patrones climáticos y apoyar decisiones preventivas desde una mirada exploratoria.

4. **Gestionar las falsas alertas mediante interpretación conjunta de métricas.**  
   El Árbol de Decisión presenta un buen recall para detectar lluvia, pero su precisión indica que todavía puede generar falsas alertas. Por esta razón, las predicciones no deben evaluarse solo por accuracy, sino también por precision, recall, F1-Score, ROC-AUC y matriz de confusión. En la práctica, esto permite interpretar las alertas con mayor criterio: un aviso de lluvia debe entenderse como una señal preventiva que requiere revisión, especialmente en zonas donde el modelo tiende a sobreestimar la ocurrencia de lluvia.

5. **Fortalecer futuras versiones con modelos no lineales para estimar cantidad de lluvia.**  
   Debido a que la Regresión Lineal presentó un `R²` bajo para `RISK_MM`, se recomienda que futuras mejoras incluyan modelos capaces de capturar relaciones no lineales, como Árboles de Regresión, Random Forest Regressor o Gradient Boosting. Esta acción permitiría mejorar la estimación de milímetros de lluvia, complementando la clasificación de ocurrencia de lluvia con una predicción más precisa de intensidad.

Estas acciones permiten conectar los resultados técnicos con decisiones concretas. De esta forma, el proyecto no se limita a entrenar modelos, sino que transforma los hallazgos en recomendaciones aplicables para monitoreo, prevención y planificación climática.

```python
# 7.1 DATAFRAME PRINCIPAL PARA POWER BI

X_full = X_final.copy()
indices_full = X_full.index

columnas_base = [
    "Date", "Location", "MinTemp", "MaxTemp", "Rainfall",
    "WindGustSpeed", "WindSpeed9am", "WindSpeed3pm",
    "Humidity9am", "Humidity3pm", "Pressure3pm",
    "RainToday", "RainTomorrow", "RISK_MM"
]

columnas_base = [col for col in columnas_base if col in df.columns]

df_powerbi_resultados = df.loc[indices_full, columnas_base].copy()

# Fecha y variables temporales
if "Date" in df_powerbi_resultados.columns:
    df_powerbi_resultados["Date"] = pd.to_datetime(df_powerbi_resultados["Date"], errors="coerce")
    df_powerbi_resultados["Anio"] = df_powerbi_resultados["Date"].dt.year
    df_powerbi_resultados["Mes"] = df_powerbi_resultados["Date"].dt.month
    df_powerbi_resultados["Nombre_Mes"] = df_powerbi_resultados["Date"].dt.month_name()

# Identificador
df_powerbi_resultados["ID_Registro"] = df_powerbi_resultados.index

# Conjunto train/test
df_powerbi_resultados["Conjunto"] = "Entrenamiento"
df_powerbi_resultados.loc[X_test.index, "Conjunto"] = "Prueba"

# Variable real de clasificación
df_powerbi_resultados["RainTomorrow_real"] = y_final.loc[indices_full].astype(int)
df_powerbi_resultados["RainTomorrow_real_label"] = np.where(
    df_powerbi_resultados["RainTomorrow_real"] == 1,
    "Sí llueve",
    "No llueve"
)

# Predicción del mejor modelo: Árbol de Decisión
df_powerbi_resultados["RainTomorrow_pred_arbol"] = modelo_arbol.predict(X_full).astype(int)
df_powerbi_resultados["RainTomorrow_pred_arbol_label"] = np.where(
    df_powerbi_resultados["RainTomorrow_pred_arbol"] == 1,
    "Predice lluvia",
    "Predice no lluvia"
)

# Probabilidad estimada de lluvia por Árbol de Decisión
if hasattr(modelo_arbol, "predict_proba"):
    df_powerbi_resultados["Probabilidad_Lluvia_Arbol"] = modelo_arbol.predict_proba(X_full)[:, 1]
else:
    df_powerbi_resultados["Probabilidad_Lluvia_Arbol"] = np.nan

# Clasificación del resultado para matriz de decisión
def clasificar_error(real, pred):
    if real == 1 and pred == 1:
        return "Verdadero Positivo"
    elif real == 0 and pred == 0:
        return "Verdadero Negativo"
    elif real == 0 and pred == 1:
        return "Falso Positivo"
    elif real == 1 and pred == 0:
        return "Falso Negativo"

df_powerbi_resultados["Tipo_Resultado_Arbol"] = df_powerbi_resultados.apply(
    lambda row: clasificar_error(row["RainTomorrow_real"], row["RainTomorrow_pred_arbol"]),
    axis=1
)

df_powerbi_resultados["Acierto_Arbol"] = (
    df_powerbi_resultados["RainTomorrow_real"] == df_powerbi_resultados["RainTomorrow_pred_arbol"]
).astype(int)

# Regresión: valores reales y predichos de RISK_MM
df_powerbi_resultados["RISK_MM_real"] = pd.to_numeric(
    df.loc[indices_full, "RISK_MM"],
    errors="coerce"
).fillna(0)

df_powerbi_resultados["RISK_MM_predicho"] = np.maximum(
    0,
    modelo_reg_lineal.predict(X_full)
)

df_powerbi_resultados["Residuo_RISK_MM"] = (
    df_powerbi_resultados["RISK_MM_real"] - df_powerbi_resultados["RISK_MM_predicho"]
)

df_powerbi_resultados["Error_Absoluto_RISK_MM"] = (
    df_powerbi_resultados["Residuo_RISK_MM"].abs()
)

# Clustering: asignación de clúster a todos los registros
X_kmeans_full = X_full[variables_kmeans].copy()
df_powerbi_resultados["Cluster_KMeans"] = kmeans_eval.predict(X_kmeans_full)

# Etiquetas dinámicas para los clústeres según tasa real de lluvia
tasas_cluster = (
    df_powerbi_resultados
    .groupby("Cluster_KMeans")["RainTomorrow_real"]
    .mean()
    .sort_values(ascending=False)
)

etiquetas_cluster = {}
nombres_etiquetas = [
    "Mayor riesgo de lluvia",
    "Riesgo intermedio",
    "Perfil más seco"
]

for i, cluster in enumerate(tasas_cluster.index):
    if i < len(nombres_etiquetas):
        etiquetas_cluster[cluster] = nombres_etiquetas[i]
    else:
        etiquetas_cluster[cluster] = f"Perfil climático {cluster}"

df_powerbi_resultados["Etiqueta_Cluster"] = df_powerbi_resultados["Cluster_KMeans"].map(etiquetas_cluster)

display(df_powerbi_resultados.head())
print("Dimensiones df_powerbi_resultados:", df_powerbi_resultados.shape)
```
